```python
import pandas as pd
import numpy as np
```

#creating dataset

```python
data ={
    "name":["sai","dev","vandana","pramodh"],
    "Age":[25,26,27,np.NaN],
    "department":["Hr","It","Sales","It"],
    "Salary":[1000,1100,1200,1300]
}
```

#data frame

```python
df = pd.DataFrame(data)
```

```python
df
```

#Filling null values with mean

```python
df["Age"]=df["Age"].fillna(round(df["Age"].mean()))
```

```python
df
```

```python
df.info()
```

#changing datatype

```python
df["Age"] = df["Age"].astype("int64")
```
