here you can see the max number as 16 and starting number is 0 here in this case scaler compress the numbers in between -2 to 2

After scaling:

Each feature will have mean = 0

Each feature will have standard deviation = 1

This is also called z-score normalization.

Why do we use Standard Scaler?

Many ML algorithms are sensitive to the scale of features:

Distance-based models (KNN, K-means, SVM, PCA) → distances would be biased if features have different scales.

Gradient descent–based models (Logistic Regression, Neural Nets) → converge faster when features are on the same scale.

Prevents features with larger numeric ranges (e.g., "salary" vs "age") from dominating the model.

```python
scaler = StandardScaler()
x_scaler = scaler.fit_transform(x)
```

```python
x_scaler
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test= train_test_split(x_scaler,y,test_size=0.2,random_state=30)
```

random_state=30

Controls the random shuffling before splitting.

Setting this makes the split reproducible (same split every time you run).

Without it, you’d get different train/test sets on each run.

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression()
```

```python
model.fit(x_train,y_train)
```

```python
model.score(x_test,y_test)
```

                                                            PCA

```python
from sklearn.decomposition import PCA
```

```python
PCA = PCA(0.95)
x_pca = PCA.fit_transform(x)
```

```python
x_pca.shape
```

```python
PCA.explained_variance_ratio_
```

After you fit PCA (pca.fit(X)), you can check pca.explained_variance_ratio_.

It gives you an array of numbers, one for each principal component.

Each number tells you how much variance (information) that component explains compared to the total variance.

```python
PCA.n_components_
```

```python
x_train_pca,x_test_pca,y_train_pca,y_test_pca= train_test_split(x_pca,y,test_size=0.2,random_state=30)
```

```python
model_pca = LogisticRegression(max_iter=1000)
model_pca.fit(x_train_pca,y_train_pca)
```

```python
model_pca.score(x_test_pca,y_test_pca)
```

                                                        N_COMPONENT

```python
from sklearn.decomposition import PCA
pca_ncomp = PCA(n_components = 45)
x_pca = pca_ncomp.fit_transform(x)
```

```python
x_pca.shape
```

```python
pca_ncomp.explained_variance_ratio_
```

```python
x_train_pca,x_test_pca,y_train_pca,y_test_pca= train_test_split(x_pca,y,test_size=0.2,random_state=30)
```

```python
model_pca = LogisticRegression(max_iter=1000)
model_pca.fit(x_train_pca,y_train_pca)
```

```python
model_pca.score(x_test_pca,y_test_pca)
```
