```python
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Housing.csv")
```

```python
df
```

```python
df1 = df[["area","price"]]
```

```python
df
```

```python
df1
```

```python
%matplotlib inline
plt.xlabel("area in sq.ft")
plt.ylabel("price in Rs.")
plt.scatter(df1["area"],df1["price"],color="green",marker=".")
```

```python
area = df1.drop('price',axis=1)
price = df1.drop('area',axis=1)
```

```python
area
```

```python
price
```

NOW CALLING THE MODEL

```python
model = linear_model.LinearRegression()
```

```python
model
```

TRAINING THE MODEL

```python
model.fit(area,price) #area is input, price is the output
```

PREDICT THE OUTPUT

```python
model.predict([[8000]])
```

```python
df1
```

prove y = mx +c. 

```python
#here i took x = 8000 
```

C VALUE

```python
c = model.intercept_
c
```

M value

```python
m = model.coef_
m
```

```python
y = m*8000 + c
y
```
