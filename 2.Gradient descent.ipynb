GD IS A OPTIMISER, ALREADY WE KNOW THE ERROR THEN WE TRY TO REDUCE THE ERROR BY USING THE GD

```python
import pandas as pd
import numpy as np
from sklearn import linear_model
import matplotlib.pyplot as plt
```

See notes for the detailed information


```python
x = np.array([1,2,3,4,5])
y = np.array([6,7,8,9,10])
```

```python
m_curr = c_curr = 0
```

```python
iterations = 310
n = len(x)
learning_rate = 0.05
```

```python
for i in range(iterations):
    y_pred = m_curr * x + c_curr
    cost = (1/n) * sum([val**2 for val in (y - y_pred)])
    md = -(2/n) * sum(x*(y-y_pred))
    cd = -(2/n) * sum(y-y_pred)
    m_curr = m_curr - learning_rate * md
    c_curr = c_curr - learning_rate * cd
    print("slope:",m_curr,"intercept: ", c_curr, "iterations:", i, "loss: ", cost) 
```
